# SQL EDA Business Case: Global Job Market Intelligence Analysis

## Business Question 4: Market Landscape Exploratory Data Analysis

## Scenario

You've just joined **TalentPulse Analytics**, a market research firm that provides competitive intelligence and workforce insights to Fortune 500 companies, staffing agencies, and HR technology vendors. Your team has acquired a large-scale job postings dataset covering global hiring activity across major employers.

The **VP of Research** has tasked you with conducting an **initial exploratory data analysis (EDA)** to help the firm understand what's in the data and identify potential insights to pitch to clients. She's specifically said:

*"We just ingested 3+ million job postings from multiple sources, and I need to understand what we're working with before we start packaging this into client deliverables. I need you to profile this dataset across multiple dimensions—geography, companies, roles, skills, and posting patterns. Give me the lay of the land so I can decide which angles are worth deeper analysis and which client segments to target."*

The VP has outlined several areas of interest, but she wants you to explore broadly rather than go deep on any single question. This is classic **exploratory analysis**—understanding distributions, identifying patterns, and surfacing anything unusual or noteworthy.

---

## Analytical Tasks

Your EDA should cover the following areas. Each task should produce **insights** (not just numbers), so think about what the data is telling you as you analyze it.

### 1. **Dataset Overview & Data Quality**
   - How many total job postings are in the dataset?
   - What is the date range of postings (earliest and most recent `date_posted`)?
   - How many unique companies are represented?
   - Are there any obvious data quality issues? (e.g., missing values in key fields like `company_name`, `job_title_standardized`, `consolidated_jds`)

### 2. **Geographic Distribution**
   - Which countries have the most job postings? (Top 10)
   - What percentage of postings are in the United States vs. international markets?
   - Are there any interesting regional patterns? (e.g., concentration in specific countries)

### 3. **Company Hiring Activity**
   - Which companies are posting the most jobs? (Top 15)
   - What is the distribution of posting volume across companies? (Are a few companies dominating, or is it evenly distributed?)
   - Calculate: How many companies posted only 1 job vs. 10+ jobs vs. 100+ jobs?

### 4. **Role & Job Title Analysis**
   - What are the most common standardized job titles? (Top 20)
   - How diverse is the role distribution? (Are there many unique roles, or is it concentrated in a few?)
   - *Bonus if you can extract patterns*: What role types appear most frequently? (e.g., "Engineer", "Manager", "Analyst")

### 5. **Skills Landscape (Tech Stack & Core Skills)**
   - What are the most frequently mentioned technologies in the `tech_stack` field? (Top 15-20)
   - What are the most in-demand core skills? (Top 15-20)
   - *Note*: These fields are stored as arrays/lists in string format—you'll need to parse them. Look for patterns like `['skill1', 'skill2']` or similar.

### 6. **Experience & Education Requirements**
   - What is the distribution of minimum experience requirements?
   - What education levels are most commonly required? (e.g., Bachelors, Masters, etc.)
   - Are there postings with no education/experience requirements specified?

### 7. **Temporal Posting Patterns**
   - Are there any noticeable trends in when jobs were posted? (e.g., posting volume by month/quarter)
   - What is the distribution of `date_posted` vs. `date_updated`? (Are most postings recent, or is this historical data?)
   - Identify the most active posting months (if the dataset spans multiple months/years)

# 2. Geographic Distribution

### Which countries have the most job postings (top 10)

```sql
-- Which countries have the most job postings? (top 10)

/* 

The countries result table returned a null value, so I had to check how
many rows contained nulls for the country field in the consolidated_jds 
column 

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;


SELECT COUNT(*) all_empty_countries
FROM 
  (
  SELECT DISTINCT explode(consolidated_jds.country) AS unique_countries
FROM challenges.challenge_tables.many_job_posts
)
WHERE 
  unique_countries IS NULL;
```

```sql
-- Which countries have the most job postings? (top 10)

/* 

This is a validation table for number of unique countries exploded.

The consolidated_jds column has arrays with keys and values,
so the country key is exploded in each cell, where each cell
contains multiple countries, because each job title is posted
in different countries.

The inner query explodes the key, and the outer query selects and filters.

 */

USE CATALOG challenges;
USE SCHEMA challenge_tables;

SELECT unique_countries
FROM 
  (
  SELECT DISTINCT explode(consolidated_jds.country) AS unique_countries
FROM challenges.challenge_tables.many_job_posts
)
WHERE 
  unique_countries IS NOT NULL
LIMIT 20;
```

```sql
-- Which countries have the most job postings? (top 10)

/* 

1. Exploding the arrays (inner query), without DISTINCT got all of the countries out from every job title (one row).
2. Every job title (row) has multiple countries, some repeated, listed in consolidated_jds.
3. After exploding, repeated countries are counted in the outer query.
4. Repeated countries are then collapsed in the outer query with
GROUP BY 

What this means is that a single row can have United States listed multiple times, hence country_count > total_rows

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;


SELECT 
  unique_countries,
  COUNT(unique_countries) AS total_country_count
FROM 
  (
  SELECT explode(consolidated_jds.country) AS unique_countries
FROM challenges.challenge_tables.many_job_posts
)
WHERE 
  unique_countries IS NOT NULL
GROUP BY 
  unique_countries
ORDER BY 
  total_country_count DESC
LIMIT 10;
```

### What percentage of postings are in US vs. International Markets?

```sql
--What percentage of postings are in the United States vs. International Markets?

/* 

This is a visual of how the ratios are computed using cross join. 
The actual calculation is the next cell after, showing percentages.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

WITH country_counts AS (
SELECT 
  unique_countries,
  COUNT(unique_countries) AS total_country_count
FROM 
  (
  SELECT explode(consolidated_jds.country) AS unique_countries
FROM challenges.challenge_tables.many_job_posts
)
WHERE 
  unique_countries IS NOT NULL
GROUP BY 
  unique_countries
ORDER BY 
  total_country_count DESC
),


total_posts AS (
  SELECT 
    SUM(total_country_count) AS all_posts
  FROM 
    country_counts
)

SELECT *
FROM country_counts
CROSS JOIN total_posts
LIMIT 10;
```

```sql
--What percentage of postings are in the United States vs. International Markets?

/* 

1. For this, we take the exploded country totals individually, and sum them because
this represents all posts within individual rows (each title/row has multiple country listings)
2. Then we cross join the unique country counts with total posts to calculate the ratios for the top 10 

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

WITH country_counts AS (
SELECT 
  unique_countries,
  COUNT(unique_countries) AS total_country_count
FROM 
  (
  SELECT explode(consolidated_jds.country) AS unique_countries
FROM challenges.challenge_tables.many_job_posts
)
WHERE 
  unique_countries IS NOT NULL
GROUP BY 
  unique_countries
ORDER BY 
  total_country_count DESC
),


total_posts AS (
  SELECT 
    SUM(total_country_count) AS all_posts
  FROM 
    country_counts
)

SELECT 
  country_counts.unique_countries,
  ROUND((country_counts.total_country_count / total_posts.all_posts) * 100, 2) AS country_ratios
FROM country_counts
CROSS JOIN total_posts
ORDER BY 
  country_ratios DESC
LIMIT 10;


```

### Are there any regional patterns?

```sql
--Are there any interesting regional patterns?

/*

This question is being pursued intuitively for the most 
distilled categorization of job types.

Since job titles are all unique (having 1 occurrence each),
the business function column is being used as the most 
distilled categorization. 

It results in 23 unique business functions, which can be used to
see if there is an observable regional pattern.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

SELECT 
  DISTINCT explode(business_function) AS unique_functions
FROM 
  challenges.challenge_tables.many_job_posts
```

```sql
--Are there any interesting regional patterns?

/*

The above function explosion is used to see
how many occurrences of each function are in the posts.

This is a higher level view of functions alone, which will
be combined with regions and roles for a more granular view.

Null function values are left alone for a clear picture of what is available.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

SELECT 
  unique_functions,
  COUNT(*) AS function_occurrence_count
FROM (
SELECT 
  explode(business_function) AS unique_functions
FROM 
  challenges.challenge_tables.many_job_posts
)
GROUP BY 
  unique_functions
ORDER BY 
  function_occurrence_count DESC;

```

```sql
--Are there any interesting regional patterns?

/*

The above business functions are now accompanied by exploded synon_locations,
which are the more specific regions hosting a job title.

Grouping by each function/region gives a count of how many times each
function is paired with a region.

Nulls are left to see a clear picture of distribution but can be cleaned.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

SELECT 
  exploded_functions,
  exploded_regions,
  COUNT(exploded_regions) AS function_region_ocurrences 
FROM (
SELECT
  explode(business_function) AS exploded_functions,
  explode(consolidated_jds.synon_location) AS exploded_regions
FROM 
  challenges.challenge_tables.many_job_posts
)
GROUP BY 
  exploded_functions,
  exploded_regions
ORDER BY 
  function_region_ocurrences DESC
LIMIT 20;
```

```sql
--Are there any interesting regional patterns?

/*

This query takes the above functions and attempts to 
reconcile empty function arrays with exploded job roles.

This creates a more granular view of regions and roles.

There are some job roles that help fulfill the uknown function e.g. IT Security Architect
would be IT function, but the function array is empty.

Heavy concentration in US and India for all functions. 

No other country appears until
entry 89, UK, with software dev engineer, 2377 entries.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

SELECT 
  exploded_job_roles,
  exploded_functions,
  exploded_regions,
  COUNT(exploded_regions) AS function_region_ocurrences 
FROM (
SELECT
  explode(draup_job_role) AS exploded_job_roles,
  explode(business_function) AS exploded_functions,
  explode(consolidated_jds.synon_location) AS exploded_regions
FROM 
  challenges.challenge_tables.many_job_posts
)
GROUP BY 
  exploded_job_roles,
  exploded_functions,
  exploded_regions
ORDER BY 
  function_region_ocurrences DESC
LIMIT 20;
```

### Distribution of posting volume across companies

```sql
--What is the distribution of posting volume across companies? (Are a few companies dominating, or is it evenly distributed?)

/* 

Calculating percent distribution again, using cross join of two CTEs.

Keeping the total post count in the table for a clear view of the calculation, transparency.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

WITH company_count AS (
SELECT 
  company_name,
  COUNT(company_id) AS company_ocurrences
FROM 
  many_job_posts
GROUP BY
  company_name
ORDER BY 
  company_ocurrences DESC
),
companies_totaled AS (
  SELECT 
    SUM(company_ocurrences) AS total_posts_all_companies
  FROM company_count
)
SELECT 
  company_name,
  company_ocurrences,
  total_posts_all_companies,
  ROUND((company_ocurrences / total_posts_all_companies) * 100, 2) AS company_percent_distr
FROM 
  company_count
CROSS JOIN 
  companies_totaled
ORDER BY 
  company_percent_distr DESC;

```

```sql
--What is the distribution of posting volume across companies? (Are a few companies dominating, or is it evenly distributed?)

/* 

Getting the spread/variation to statistically visualize 
the data skew.

Variation coefficient of 75% shows high data spread/inconsistency, 
which can be explained by the dominance of the top two companies
Amazon and Accenture.

*/
--What is the distribution of posting volume across companies? (Are a few companies dominating, or is it evenly distributed?)

/* 

Calculating percent distribution again, using cross join of two CTEs.

Keeping the total post count in the table for a clear view of the calculation, transparency.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

WITH company_count AS (
SELECT 
  company_name,
  COUNT(company_id) AS company_ocurrences
FROM 
  many_job_posts
GROUP BY
  company_name
ORDER BY 
  company_ocurrences DESC
),
companies_totaled AS (
  SELECT 
    SUM(company_ocurrences) AS total_posts_all_companies
  FROM company_count
),
spread_stats AS(
SELECT 
  company_name,
  company_ocurrences,
  AVG(company_ocurrences) OVER() AS company_ocurrence_mean,
  ROUND(stddev_pop(company_ocurrences) OVER(), 2) AS company_occurrence_std
FROM 
  company_count
ORDER BY 
  company_ocurrences DESC
)
SELECT 
  company_name,
  company_ocurrence_mean,
  company_occurrence_std,
  ROUND(company_occurrence_std / company_ocurrence_mean, 2) AS variation_coefficient,
  ROUND((company_ocurrences - company_ocurrence_mean) / company_occurrence_std, 2) AS company_z_score
FROM 
  spread_stats;





```

```sql
-- Calculate: How many companies posted only 1 job vs. 10+ jobs vs. 100+ jobs?

/* 

Using the above table, just reversing the order to see least amount of posts by companies.

No companies only posted 1 or 10 jobs. The lowest posting amount by Sony is 14,654.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;

WITH company_count AS (
SELECT 
  company_name,
  COUNT(company_id) AS company_ocurrences
FROM 
  many_job_posts
GROUP BY
  company_name
ORDER BY 
  company_ocurrences DESC
),
companies_totaled AS (
  SELECT 
    SUM(company_ocurrences) AS total_posts_all_companies
  FROM company_count
)
SELECT 
  company_name,
  company_ocurrences,
  total_posts_all_companies,
  ROUND((company_ocurrences / total_posts_all_companies) * 100, 2) AS company_percent_distr
FROM 
  company_count
CROSS JOIN 
  companies_totaled
ORDER BY 
  company_percent_distr ASC;

```

# 6. Experience & Education Requirements

### What is the distribution of minimum experience requirements?

```sql
--What is the distribution of minimum experience requirements?


/* 

This is a normalization query to clean the years of experience column for analysis.

Experience is given in months, years, days, and sometimes unclear strings.

This normalization query brings everything under years with the CASE statement,
casting strings anything without months or days to the solid number provided.

Next, it will be determined how many rows are actually normalized, and can be
analyzed.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;


SELECT
  minimum_index_experience,
  CASE 
    WHEN lower(minimum_index_experience) LIKE '%month%' 
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 12, 1)
    WHEN lower(minimum_index_experience) LIKE '%day%'
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 365, 1)
    ELSE CAST(regexp_substr(minimum_index_experience, '\\d+') AS INT)
  END AS years_exp_normalized
FROM 
  many_job_posts
WHERE 
  minimum_index_experience IS NOT NULL
LIMIT 20;
```

```sql
--What is the distribution of minimum experience requirements?


/* 

This query shows a sample of the strings that cannot be normalized, which are included in the null 
values that will be excluded.

The percentage of usable values are presented next.

*/


USE CATALOG challenges;
USE SCHEMA challenge_tables;

SELECT 
  minimum_index_experience,
  years_exp_normalized
FROM (
SELECT
  minimum_index_experience,
  CASE 
    WHEN lower(minimum_index_experience) LIKE '%month%' 
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 12, 1)
    WHEN lower(minimum_index_experience) LIKE '%day%'
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 365, 1)
    ELSE CAST(regexp_substr(minimum_index_experience, '\\d+') AS INT)
  END AS years_exp_normalized
FROM 
  many_job_posts
WHERE 
  minimum_index_experience IS NOT NULL
)
WHERE 
  years_exp_normalized IS NULL
LIMIT 20;
```

```sql
--What is the distribution of minimum experience requirements?


/* 

Total usable experience rows and total null experience rows are divided by total 
row count to find out what percentage of the data will be analyzed for 
distribution metrics.

The row count is placed at the end for transparency.

*/


USE CATALOG challenges;
USE SCHEMA challenge_tables;

WITH years_exp_norm AS (
SELECT
  minimum_index_experience,
  CASE 
    WHEN lower(minimum_index_experience) LIKE '%month%' 
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 12, 1)
    WHEN lower(minimum_index_experience) LIKE '%day%'
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 365, 1)
    ELSE CAST(regexp_substr(minimum_index_experience, '\\d+') AS INT)
  END AS years_exp_normalized
FROM 
  many_job_posts
),
total_rows AS (
  SELECT 
    COUNT(*) AS row_count
  FROM 
    years_exp_norm
),
exp_null_count AS(
SELECT 
  COUNT(*) AS all_null_exp
FROM 
  years_exp_norm
WHERE 
  years_exp_normalized IS NULL
),
not_null_exp_rows AS (
  SELECT 
    COUNT(*) AS not_null_experience
  FROM 
    years_exp_norm
  WHERE 
    years_exp_normalized IS NOT NULL
)
SELECT 
  all_null_exp,
  ROUND((all_null_exp / row_count) * 100, 2) AS null_exper_perc_drop,
  not_null_experience,
  ROUND((not_null_experience / row_count) * 100, 2) AS normalized_exper_perc_keep,
  row_count AS row_transparency
FROM 
  exp_null_count
CROSS JOIN
  not_null_exp_rows
CROSS JOIN 
  total_rows;
```

```sql

--What is the distribution of minimum experience requirements?


/* 

Continuing to work with the experience column, the normalized values 
still have some messy issues.

Values greater than 30 years begin to become ambiguous and chaotically distinct, 
making filtering too granular.

The most viable approach would be exclude rows that exceed the 30 year
threshold.

*/


USE CATALOG challenges;
USE SCHEMA challenge_tables;


WITH usable_normalized_experience AS (
SELECT
  job_title_standardized,
  minimum_index_experience,
  CASE 
    WHEN lower(minimum_index_experience) LIKE '%month%' 
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 12, 1)
    WHEN lower(minimum_index_experience) LIKE '%day%'
      THEN ROUND(CAST(regexp_substr(minimum_index_experience, '\\d+') AS DOUBLE) / 365, 1)
    ELSE CAST(regexp_substr(minimum_index_experience, '\\d+') AS INT)
  END AS years_exp_normalized
FROM 
  many_job_posts
WHERE 
  minimum_index_experience IS NOT NULL
)

SELECT
  job_title_standardized,
  minimum_index_experience,
  years_exp_normalized
FROM usable_normalized_experience
GROUP BY 
  job_title_standardized,
  minimum_index_experience,
  years_exp_normalized
ORDER BY 
 years_exp_normalized DESC
LIMIT 20;

```

### Are there any noticeable trends in when jobs were posted? (e.g., posting volume by month/quarter)


```sql
-- Are there any noticeable trends in when jobs were posted? (e.g., posting volume by month/quarter)


/* 

This table aggregates posting dates by year, month, and quarter, and 
counts the occurrences of each set of attributes.

This provides an overview of which month had
the most job postings in the dataset, by its
year and quarter.

*/

USE CATALOG challenges;
USE SCHEMA challenge_tables;


WITH month_counts AS (
SELECT
  date_posted,
  COUNT(date_format(date_posted, 'MMM')) AS month_count
FROM 
  many_job_posts
GROUP BY
  date_posted
)
SELECT 
  year(date_posted) AS year_posted,
  date_format(date_posted, 'MMM') AS month_names, 
  date_part('QUARTER', date_posted) AS date_quarter,
  month_count
FROM 
  month_counts
ORDER BY 
  month_count DESC
LIMIT 20;
```

```sql
-- Are there any noticeable trends in when jobs were posted? (e.g., posting volume by month/quarter)


/* 

This is an aggregation focused on counting posts
by year. 

This shows which year had the most posts in the entire dataset
by month and quarter.

*/


USE CATALOG challenges;
USE SCHEMA challenge_tables;


WITH month_counts AS (
SELECT
  date_posted,
  COUNT(date_format(date_posted, 'MMM')) AS month_count
FROM 
  many_job_posts
GROUP BY
  date_posted
),
date_agg AS (
SELECT 
  year(date_posted) AS year_posted,
  date_format(date_posted, 'MMM') AS month_names, 
  date_part('QUARTER', date_posted) AS date_quarter,
  month_count
FROM 
  month_counts
ORDER BY 
  month_count DESC
)

SELECT 
  year_posted,
  month_names,
  date_quarter,
  COUNT(year_posted) AS year_count
FROM 
  date_agg
GROUP BY 
  year_posted,
  month_names,
  date_quarter
ORDER BY 
  year_count DESC
LIMIT 20;
```

```sql
-- Are there any noticeable trends in when jobs were posted? (e.g., posting volume by month/quarter)


/* 

The percent of the job posts that come from each year.

2022 makes up the majority of the original job posts,
while 2025 and 2021 are next up.

This suggests the majority of the dataset is historical,
and the second largest portion is newer jobs.

*/


USE CATALOG challenges;
USE SCHEMA challenge_tables;


WITH month_counts AS (
SELECT
  date_posted,
  COUNT(date_format(date_posted, 'MMM')) AS month_count
FROM 
  many_job_posts
GROUP BY
  date_posted
),
date_agg AS (
SELECT 
  year(date_posted) AS year_posted,
  date_format(date_posted, 'MMM') AS month_names, 
  date_part('QUARTER', date_posted) AS date_quarter,
  month_count
FROM 
  month_counts
ORDER BY 
  month_count DESC
),
all_posts AS(
  SELECT 
    COUNT(*) AS total_post_count
  FROM 
    many_job_posts
),
post_year_distr_table AS (
SELECT 
  year_posted,
  COUNT(year_posted) AS year_ocurrences,
  total_post_count
FROM 
  date_agg
CROSS JOIN 
  all_posts
GROUP BY 
  year_posted,
  total_post_count
ORDER BY 
  year_ocurrences DESC
)

SELECT 
  year_posted,
  year_ocurrences,
  total_post_count,
  ROUND((year_ocurrences / total_post_count) * 100, 2) AS year_post_distr_perc
FROM 
  post_year_distr_table
GROUP BY 
  year_posted,
  total_post_count,
  year_ocurrences
ORDER BY 
  year_post_distr_perc DESC;
```

### What is the distribution of `date_posted` vs. `date_updated`? (Are most postings recent, or is this historical data?)


```sql

-- What is the distribution of `date_posted` vs. `date_updated`? (Are most postings recent, or is this historical data?)

/* 

This shows 2025, 2024, 2022 as the years with the most posted and updated records 
in the dataset.

This suggests that newer records with updates make up the majority of 
the dataset.

To reconcile the previous table, original posts mostly come from 2022,
but when it comes to records with updates, newer job posts with 
updates are more plentiful than original, 2022 posts.

*/


USE CATALOG challenges;
USE SCHEMA challenge_tables;

WITH all_dates AS (
SELECT 
  CAST(date_posted AS DATE) AS all_post_dates,
  CAST(date_updated AS DATE) AS all_update_dates
FROM 
  many_job_posts
),
post_update_years AS(
SELECT 
  year(all_post_dates) AS all_post_years,
  year(all_update_dates) AS all_update_years 
FROM 
  all_dates
)

SELECT 
  all_post_years,
  all_update_years,
  COUNT(all_post_years) AS year_post_count
FROM 
  post_update_years
GROUP BY 
  all_post_years,
  all_update_years
ORDER BY 
  year_post_count DESC;
```
